```python
def clean_query(query):
    query = query.lower()

    # Remove common instruction phrases
    phrases_to_remove = [
        "tell me about",
        "explain about",
        "explain",
        "describe",
        "give details about",
        "what is",
        "provide details about",
        "information about"
    ]

    for phrase in phrases_to_remove:
        if phrase in query:
            query = query.replace(phrase, "")

    return query.strip()
```

```python
import requests
import os
from langchain_community.document_loaders import PyPDFLoader

def ask_question(query):

    cleaned_query = clean_query(query)

    print("🔎 Cleaned Query:", cleaned_query)

    results = retriever.invoke(cleaned_query)[:3]


    context = "\n".join([doc.page_content for doc in results])

    print("🧠 Sending to Gemini...")

    prompt = f"""
You are a professional IT Policy Assistant.

Use the context if relevant.
Be clear and structured.

Context:
{context}

Question:
{query}

Answer:
"""

    API_KEY = os.environ.get("GOOGLE_API_KEY")
    url = f"https://generativelanguage.googleapis.com/v1/models/gemini-2.5-flash:generateContent?key={API_KEY}"

    data = {
        "contents": [
            {"parts": [{"text": prompt}]}
        ]
    }

    response = requests.post(
        url,
        headers={"Content-Type": "application/json"},
        json=data,
        timeout=20
    )

    if response.status_code != 200:
        print(f"Error: API request failed with status code {response.status_code}")
        print(response.text)
        return "Sorry, I couldn't get an answer from the API due to a request error."

    result = response.json()

    if "candidates" in result and len(result["candidates"]) > 0:
        return result["candidates"][0]["content"]["parts"][0]["text"]
    else:
        print("Error: No candidates or unexpected response structure found in API response.")
        print(result) # Print the full response for debugging
        return "Sorry, I couldn't generate an answer based on the retrieved context or there was an issue with the API response."
```

```python
question = "introduction"

print("\n⏳ Generating answer...\n")
answer = ask_question(question)
print("✅ Answer:\n")
print(answer)
```

**Reasoning**:

```python
import os
import pandas as pd
from langchain_community.document_loaders import PyPDFLoader, UnstructuredWordDocumentLoader
from langchain_core.documents import Document

documents = []

def get_files_from_directory(directory):
    file_paths = []
    for root, _, files in os.walk(directory):
        for file in files:
            file_paths.append(os.path.join(root, file))
    return file_paths

# Get files from the current directory
current_dir_files = get_files_from_directory('.')

# Get files from Google Drive
drive_dir = '/content/drive/MyDrive/'
if os.path.exists(drive_dir):
    drive_files = get_files_from_directory(drive_dir)
else:
    drive_files = []
    print(f"Warning: Google Drive directory '{drive_dir}' not found. Skipping Drive files.")

# Combine all file paths
all_files = current_dir_files + drive_files

# Filter for supported document types and load them
for file_path in all_files:
    if file_path.endswith(".pdf"):
        try:
            loader = PyPDFLoader(file_path)
            documents.extend(loader.load())
        except Exception as e:
            print(f"Error loading PDF {file_path}: {e}")

    elif file_path.endswith(".docx"):
        try:
            loader = UnstructuredWordDocumentLoader(file_path)
            documents.extend(loader.load())
        except Exception as e:
            print(f"Error loading DOCX {file_path}: {e}")

    elif file_path.endswith(".xlsx"):
        try:
            df = pd.read_excel(file_path)
            documents.append(
                Document(
                    page_content=df.to_string(),
                    metadata={"source": file_path}
                )
            )
        except Exception as e:
            print(f"Error loading XLSX {file_path}: {e}")

    elif file_path.endswith(".csv"):
        try:
            df = pd.read_csv(file_path)
            documents.append(
                Document(
                    page_content=df.to_string(),
                    metadata={"source": file_path}
                )
            )
        except Exception as e:
            print(f"Error loading CSV {file_path}: {e}")

    elif file_path.endswith(".txt"):
        try:
            with open(file_path, "r", encoding="utf-8") as f:
                documents.append(
                    Document(
                        page_content=f.read(),
                        metadata={"source": file_path}
                    )
                )
        except Exception as e:
            print(f"Error loading TXT {file_path}: {e}")

print("Documents loaded:", len(documents))
```

## Identify the problematic file

### Subtask:
Examine the files in your Google Drive and current directory, especially any large `.csv` or `.xlsx` files, to see if their size or content might be causing the `df.to_string()` operation to hang or become very slow.


**Reasoning**:
To identify potentially problematic large files, I will create a function to gather file paths and sizes for CSV and XLSX files, apply it to both the current directory and Google Drive, and then display the sorted results in a DataFrame.



```python
import os
import pandas as pd

def get_file_info(directory):
    file_info_list = []
    for root, _, files in os.walk(directory):
        for file in files:
            file_path = os.path.join(root, file)
            if file.endswith(('.csv', '.xlsx')):
                try:
                    file_size_bytes = os.path.getsize(file_path)
                    file_size_mb = file_size_bytes / (1024 * 1024)  # Convert to MB
                    file_info_list.append({"File Path": file_path, "Size (MB)": file_size_mb})
                except Exception as e:
                    print(f"Error getting size for {file_path}: {e}")
    return file_info_list

# Get files from the current directory
current_dir_info = get_file_info('.')

# Get files from Google Drive
drive_dir = '/content/drive/MyDrive/'
drive_info = []
if os.path.exists(drive_dir):
    drive_info = get_file_info(drive_dir)
else:
    print(f"Warning: Google Drive directory '{drive_dir}' not found. Skipping Drive file scanning.")

# Combine all file information
all_file_info = current_dir_info + drive_info

# Convert to DataFrame and sort
if all_file_info:
    df_file_sizes = pd.DataFrame(all_file_info)
    df_file_sizes = df_file_sizes.sort_values(by='Size (MB)', ascending=False).reset_index(drop=True)
    print("File sizes of .csv and .xlsx files (sorted by size):")
    print(df_file_sizes)
else:
    print("No .csv or .xlsx files found in the specified directories.")
```

**Reasoning**:
To prevent resource exhaustion when converting large dataframes to strings, I will modify the document loading logic to wrap `df.to_string()` in a `try-except` block and limit the number of rows and column width for `.csv` and `.xlsx` files.



```python
import os
import pandas as pd
from langchain_community.document_loaders import PyPDFLoader, UnstructuredWordDocumentLoader
from langchain_core.documents import Document

documents = []

def get_files_from_directory(directory):
    file_paths = []
    for root, _, files in os.walk(directory):
        for file in files:
            file_paths.append(os.path.join(root, file))
    return file_paths

# Get files from the current directory
current_dir_files = get_files_from_directory('.')

# Get files from Google Drive
drive_dir = '/content/drive/MyDrive/'
if os.path.exists(drive_dir):
    drive_files = get_files_from_directory(drive_dir)
else:
    drive_files = []
    print(f"Warning: Google Drive directory '{drive_dir}' not found. Skipping Drive files.")

# Combine all file paths
all_files = current_dir_files + drive_files

# Filter for supported document types and load them
for file_path in all_files:
    if file_path.endswith(".pdf"):
        # Explicitly skip known problematic encrypted PDF
        if "EAadhaar_MAHIESHWAR.pdf" in file_path:
            print(f"Skipping known encrypted PDF: {file_path}")
            continue # Skip to the next file
        try:
            loader = PyPDFLoader(file_path)
            documents.extend(loader.load())
        except Exception as e:
            print(f"Error loading PDF {file_path}: {e}")

    elif file_path.endswith(".docx"):
        try:
            loader = UnstructuredWordDocumentLoader(file_path)
            documents.extend(loader.load())
        except Exception as e:
            print(f"Error loading DOCX {file_path}: {e}")

    elif file_path.endswith(".xlsx"):
        try:
            df = pd.read_excel(file_path)
            # Limit rows and column width for large files to prevent memory issues
            text = df.head(100).to_string(max_colwidth=100)
            documents.append(
                Document(
                    page_content=text,
                    metadata={"source": file_path}
                )
            )
        except Exception as e:
            print(f"Error loading XLSX {file_path}: {e}")

    elif file_path.endswith(".csv"):
        try:
            df = pd.read_csv(file_path)
            # Limit rows and column width for large files to prevent memory issues
            text = df.head(100).to_string(max_colwidth=100)
            documents.append(
                Document(
                    page_content=text,
                    metadata={"source": file_path}
                )
            )
        except Exception as e:
            print(f"Error loading CSV {file_path}: {e}")

    elif file_path.endswith(".txt"):
        try:
            with open(file_path, "r", encoding="utf-8") as f:
                documents.append(
                    Document(
                        page_content=f.read(),
                        metadata={"source": file_path}
                    )
                )
        except Exception as e:
            print(f"Error loading TXT {file_path}: {e}")

print("Documents loaded:", len(documents))
```
